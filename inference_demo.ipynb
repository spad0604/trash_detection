```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt

# 1. Load mô hình đã train (best_model.pt)
model_path = r'D:\Trash Detection\YOLO-Waste-Detection-main\YOLO-Waste-Detection-main\best_model.pt'
model = YOLO(model_path)
print("Đã load model thành công!")
```

```python
# 2. Dự đoán trên một bức ảnh
image_path = r'D:\Trash Detection\YOLO-Waste-Detection-main\YOLO-Waste-Detection-main\Image\image_phone.jpeg'

try:
    # Gọi hàm predict với confidence threshold là 0.5
    results = model.predict(source=image_path, conf=0.5)
    
    # Lấy mảng ảnh chứa bounding box kết quả
    res_img = results[0].plot()
    
    # Convert BGR (chuẩn của OpenCV) sang RGB (chuẩn của Matplotlib)
    res_img_rgb = cv2.cvtColor(res_img, cv2.COLOR_BGR2RGB)
    
    # 3. Hiển thị ảnh
    plt.figure(figsize=(10, 10))
    plt.imshow(res_img_rgb)
    plt.axis('off')
    plt.show()
    
    # In ra các class phát hiện được cùng độ tin cậy
    print("\n--- Kết quả phát hiện ---")
    for box in results[0].boxes:
        class_id = int(box.cls[0])
        conf = float(box.conf[0])
        class_name = model.names[class_id]
        print(f"- {class_name}: {conf:.2f}")
except Exception as e:
    print("Lỗi khi dự đoán:", e)
```
